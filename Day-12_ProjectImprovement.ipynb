##Load Dataset

```python
# 1. Import Pandas
import pandas as pd

# 2. Load a CSV dataset
df = pd.read_csv("student_sales_data.csv") 
```

```python
# 3. View rows
print(df.head())       # first 5 rows
print(df.tail())       # last 5 rows
# 4. View columns
print(df.columns)
```

```python
# 5. View dataset information
print(df.info())       # data types, non-null counts
print(df.shape)        # (rows, columns)
print(df.describe())
```

##Clean Dataset

```python
# 1. Handle missing values
print(df.isnull().sum())          # check missing values per column
df = df.dropna()                  # option 1: remove rows with missing values
# df = df.fillna(0) 
```

```python
# 2. Remove duplicate records
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
```

```python
# 3. Correct data types
print(df.dtypes)                  # check current data types
df["OrderID"] = df["OrderID"].astype(int)
df["Price"] = df["Price"].astype(float)
```

```python
# Final check
print(df.info())
print(df.shape)
```

##Data Filtering

```python
# 1. Filter rows (based on a condition)
filtered_df = df[df["Quantity"] > 3]          # example: rows where Quantity > 3
print(filtered_df)
```

```python
# 2. Select specific columns
selected_cols = df[["OrderID", "Quantity", "Price"]]   # replace with your actual column names
print(selected_cols)
```

```python
# 3. Sort the dataset
sorted_df = df.sort_values(by="Price", ascending=False)   # sort by Price, highest first
print(sorted_df)
```

```python
# Save the filtered dataset (optional, but good practice for "Filtered dataset created")
filtered_df.to_csv("filtered_dataset.csv", index=False)
```

##Data Analysis

```python
# 1. Total
total_price = df["Price"].sum()
total_quantity = df["Quantity"].sum()
print("Total Price:", total_price)
print("Total Quantity:", total_quantity)
```

```python
# 2. Average
avg_price = df["Price"].mean()
avg_quantity = df["Quantity"].mean()
print("Average Price:", avg_price)
print("Average Quantity:", avg_quantity)
```

```python
# 3. Minimum
min_price = df["Price"].min()
min_quantity = df["Quantity"].min()
print("Minimum Price:", min_price)
print("Minimum Quantity:", min_quantity)
```

```python
# 4. Maximum
max_price = df["Price"].max()
max_quantity = df["Quantity"].max()
print("Maximum Price:", max_price)
print("Maximum Quantity:", max_quantity)
```

```python
# 5. Count
count_orders = df["OrderID"].count()
print("Total Orders (Count):", count_orders)

```
